# 1. De OLAP a Machine Learning
El cubo OLAP nos entregó el reporte perfecto para humanos, pero inútil para una Red Neuronal. Nuestro trabajo es aplicar *Feature Engineering* para convertir este reporte en la Matriz Matemática $\mathbf{X}$.

```python
import pandas as pd

# Simulamos el resultado extraído de nuestro Cubo OLAP después de un Slice/Dice
cubo_exportado = pd.DataFrame({
    'id_cliente': [101, 102, 103, 104, 105],
    'ciudad': ['Ensenada', 'Tijuana', 'Tijuana', 'Mexicali', 'Ensenada'],
    'tipo_suscripcion': ['Basica', 'Premium', 'Estandar', 'Basica', 'Premium'],
    'gasto_mensual': [250.50, 890.00, 450.00, 210.00, 950.00]
})

print("--- DATOS DEL CUBO (Legibles para humanos, prohibidos para la IA) ---")
display(cubo_exportado)
```

### 2. Codificación Ordinal (Label Encoding)
La columna `tipo_suscripcion` tiene un orden lógico de negocio: Basica < Estandar < Premium.
Si usamos un mapeo directo, el algoritmo entenderá que el salto de Básica a Premium requiere subir de nivel geométrico.

```python
jerarquia_suscripcion = {"Basica": 1, "Estandar":2, "Premium":3}
cubo_exportado['suscripcion_encoded'] = cubo_exportado['tipo_suscripcion'].map(jerarquia_suscripcion)

print("--- DESPUÉS DEL LABEL ENCODING ---")
display(cubo_exportado[['tipo_suscripcion', 'suscripcion_encoded']])
```

### 3. One-Hot Encoding (Expansión de Dimensiones)
La columna `ciudad` es nominal. No podemos decirle al modelo que Ensenada es 1 y Mexicali es 3. Debemos expandir la dimensión espacial usando `pd.get_dummies()`.

```python
matriz_X = pd.get_dummies(cubo_exportado, columns=["ciudad"], dtype=int)
matriz_X = matriz_X.drop(columns=['id_cliente',"tipo_suscripcion"])

print("--- MATRIZ X DEFINITIVA PARA ENTRENAMIENTO ---")
display(matriz_X)
```
