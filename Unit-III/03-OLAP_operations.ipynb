# Laboratorio: Código Rojo en el Data Warehouse 🚨

Son los Ingenieros de Datos de "AutoMetrics". El script ETL falló. Su misión es arreglar la tubería de datos y luego ejecutar consultas OLAP para el equipo de negocios.

### Fase 1: OLTP (Extracción)
Simulamos la ingesta del sistema transaccional. Hay un error oculto en los salarios (`-999`).

```python
import pandas as pd
import numpy as np

# EXTRACT
datos_crudos = pd.DataFrame({
    'id_empleado': [1, 2, 3, 4, 5],
    'nombre_ciudad': ['Ensenada', 'Tijuana', 'Ensenada', 'Mexicali', 'Tijuana'],
    'departamento': ['Ventas', 'IT', 'Ventas', 'IT', 'Ventas'],
    'salario_mensual': [25000, 30000, -999, 28000, 32000],
    'comentario_desempeno': ["Excelente", "Impuntual", "Buen trabajo", "Cumple", "Sobresaliente"]
})
datos_crudos.head()
```

```python
# RETO 1: Limpia el valor corrupto de -999 antes de que contamine el almacén.
# TU CÓDIGO AQUÍ:
datos_crudos.loc[datos_crudos['salario_mensual'] < 0, 'salario_mensual'] = datos_crudos['salario_mensual'].median()
datos_crudos.head()
```

### Fase 2: ETL y Esquema Estrella (Transform & Load)
El ingeniero anterior provocó una Explosión Cartesiana al crear la dimensión y dejó una columna de texto gigante que destruirá la memoria RAM.

```python
# TRANSFORM
# RETO 2: Arregla la creación de la dimensión agregando el método para eliminar duplicados.
# dim_ciudad = datos_crudos[['nombre_ciudad']].copy() # <-- Arregla esta línea
dim_ciudad = datos_crudos[['nombre_ciudad']].drop_duplicates().reset_index(drop=True) # <-- Arregla esta línea
dim_ciudad['id_ciudad'] = dim_ciudad.index + 1

print(dim_ciudad.head())

dim_depto = datos_crudos[['departamento']].drop_duplicates().reset_index(drop=True)
dim_depto['id_depto'] = dim_depto.index + 1

print(dim_depto.head())

# Hacemos Merge para obtener los IDs
df_temp = datos_crudos.merge(dim_ciudad, on='nombre_ciudad', how='left')
df_temp = df_temp.merge(dim_depto, on='departamento', how='left')
df_temp.head()
```

```python
# RETO 3: Construye la Tabla de Hechos aislando solo las llaves foráneas y el salario. 
# (Deja fuera el comentario de desempeño)
fact_rendimiento = df_temp[["id_ciudad", "id_depto", "salario_mensual"]] # <-- Arregla esta línea

# LOAD (La matriz KDD se consolida en memoria)
print("Data Warehouse Cargado.")
fact_rendimiento.head()
```

### Fase 3: Operaciones del Cubo OLAP
El almacén está listo. Ahora cruzamos hechos y dimensiones para responder preguntas de negocio usando operaciones OLAP.

```python
# Reconstruimos el Cubo OLAP uniendo las tablas
cubo_olap = fact_rendimiento.merge(dim_ciudad, on='id_ciudad').merge(dim_depto, on='id_depto')
cubo_olap
```

```python


# 1. ROLL-UP (Menos detalle): Salario total por Departamento
print("--- ROLL-UP ---")
print(cubo_olap.groupby('departamento')['salario_mensual'].sum())
```

```python
# 2. DRILL-DOWN (Más detalle): Salario total por Departamento y Ciudad
print("\n--- DRILL-DOWN ---")
print(cubo_olap.groupby(['departamento', 'nombre_ciudad'])['salario_mensual'].sum())
```

```python
# RETO 4 (SLICE): Escribe una consulta que filtre el cubo SOLO para la ciudad de 'Tijuana'
slice_tijuana = cubo_olap[cubo_olap["nombre_ciudad"] == "Tijuana"]
print("\n--- SLICE ---")
print(slice_tijuana)
```

```python
# RETO 5 (DICE): Escribe una consulta que filtre el cubo para 'Ventas' EN 'Ensenada'
dice_ventas_ens = cubo_olap[(cubo_olap["nombre_ciudad"] == "Ensenada") & (cubo_olap["departamento"] == "Ventas")]
print("\n--- DICE ---")
print(dice_ventas_ens)
```
