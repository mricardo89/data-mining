```python
import pandas as pd
import numpy as np

# Reemplaza con la URL de tu repositorio
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)
```

```python
df['Rating'].mean()
```

```python
def mean_custom(values):
    accumulate = 0
    for val in values:
        accumulate += val

    return accumulate/len(values)

mean_custom(df['Rating'].dropna())
```
